## 问题描述

您需要一个能够根据房屋面积预测房价的模型。

我们沿用上一次实验中的两个数据点：一套 1000 平方英尺的房屋售价为 30 万美元，另一套 2000 平方英尺的房屋售价为 50 万美元。


| 面积（千平方英尺） | 价格（千美元） |
|-----------|---------|
| 1         | 300     |
| 2         | 500     |


```python
x_train = np.array([1.0, 2.0]) # 面积（千平方英尺）
y_train = np.array([300.0, 500.0]) # 价格（千美元）

def cost_compute(x, w, b, y):
    m = x.shape[0]
    cost_sum = 0
    for i in range(m):
        f_wb = x[i] * w + b
        cost = (f_wb - y)**2
        cost_sum += cost
    cost_sum *= 1/(2*m)
    return cost_sum

x_train = np.array([1.0, 1.7, 2.0, 2.5, 3.0, 3.2])
y_train = np.array([250, 300, 480, 430, 630, 730,])

plt.close('all')
fig, ax, dyn_items = plt_stationary(x_train, y_train)
updater = plt_update_onclick(fig, ax, x_train, y_train, dyn_items)
```
