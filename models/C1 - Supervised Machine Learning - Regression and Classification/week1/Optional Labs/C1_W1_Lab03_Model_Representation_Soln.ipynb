现在，我们来计算两个数据点的 $f_{w,b}(x^{(i)})$
 值。你可以针对每个数据点显式写出表达式：
对于 $x^{(0)}$, `f_wb = w * x[0] + b`
对于 $x^{(1)}$, `f_wb = w * x[1] + b`
当数据点数量很多时，这样写会十分繁琐且重复。因此，可以使用 for 循环计算函数输出，如下方的 compute_model_output 函数所示。
> **注意**:参数说明（ndarray (m,)）代表形状为 (m,) 的 NumPy 多维数组；（scalar）代表无维度、仅表示大小的标量。
>
> **注意**:np.zero (n) 会返回一个包含n个元素的一维 NumPy 数组。

```python
def compute_model_output(x, w, b):
    """
    计算线性模型的预测结果
    参数：
        x (ndarray (m,))：输入数据，共 m 个样本
        w,b (scalar) ：模型参数
    返回值：
        y (ndarray (m,))：目标预测值
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

现在让我们调用 compute_model_output 函数并绘制输出结果。

```python
w = 100
b = 100

tmp_f_wb = compute_model_output(x_train, w, b)

# 绘制我们的模型预测结果
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')

# 绘制数据点
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# 设置标题
plt.title("Housing Prices")
# 设置 y 轴标签
plt.ylabel('Price (in 1000s of dollars)')
# 设置 x 轴标签
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```
