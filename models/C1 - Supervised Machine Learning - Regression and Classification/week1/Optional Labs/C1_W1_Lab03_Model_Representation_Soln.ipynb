```python
# 模型展示方案

# 先导入numpy库（科学计算库）与matplotlib（绘图库）
# matplotlib.pyplot 是 matplotlib 库的一个子模块，旨在提供简单易用的绘图功能。
# 它包含了一系列函数，用户可以通过这些函数快速创建和定制图表。
# 常用的图表类型包括线图、散点图、柱状图、饼图等。使用时，通常会将其导入为 plt，以便于调用其函数
import numpy as np
import matplotlib.pyplot as plt
```

```python
# mplstyle 文件是 Matplotlib 的样式文件，以 .mplstyle 为后缀，用于定义图表的视觉风格，包括颜色、线条样式、字体、背景、网格等属性
# 每个文件本质上是一个 键-值对的配置集合，类似于 Matplotlib 的 rcParams 配置，可以在不同机器或项目中保持样式一致
plt.style.use('./deeplearning.mplstyle')
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
```

```python
# NumPy 数组有一个 `.shape` 参数。`x_train.shape` 返回一个 Python 元组，每个维度对应一个条目。
# `x_train.shape[0]` 是数组的长度和样本数量
print(f"x_train.shape = {x_train.shape}")
print(f"x_train.shape[0] = {x_train.shape[0]}")
```

```python
# 使用 matplotlib 库中的 scatter() 函数绘制
# 函数参数 x和y将数据作为散点图的X，y轴坐标 marker和c将点显示为红色十字（默认值为蓝色圆点）
plt.scatter(x=x_train, y=y_train, marker='x', c='r')
# 设置标题
plt.title("Housing Prices")
# 设置y轴标签
plt.ylabel("Price (in 1000s of dollars)")
# 设置x轴标签
plt.xlabel("Size (1000 sqft)")
# 展示图像
plt.show()
```
