```python
import matplotlib.pyplot as plt
import numpy as np

# 1. 应用样式
plt.style.use('./deeplearning.mplstyle')

# 2. 【关键修复】恢复刻度线大小，确保标签有锚点
plt.rcParams['xtick.major.size'] = 3.5
plt.rcParams['ytick.major.size'] = 3.5
plt.rcParams['xtick.minor.size'] = 2.0
plt.rcParams['ytick.minor.size'] = 2.0

# 3. 【关键修复】确保刻度标签颜色可见（防止与背景同色）
plt.rcParams['xtick.color'] = 'black'
plt.rcParams['ytick.color'] = 'black'
plt.rcParams['axes.labelcolor'] = 'black' # 轴标题颜色
plt.rcParams['text.color'] = 'black'      # 全局文字颜色

# 4. 确保字体支持（防止因字体缺失导致空白）
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'DejaVu Sans', 'Arial']
plt.rcParams['axes.unicode_minus'] = False

# 5. 绘图示例
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])

plt.scatter(x=x_train, y=y_train, marker='x', c='r', label='Training Data')
plt.title("Housing Prices")
plt.xlabel("Size (1000 sqft)")
plt.ylabel("Price (in 1000s of dollars)")

# 6. 显式设置刻度标签（可选，用于测试是否显示）
# plt.xticks([1.0, 2.0], ['1.0', '2.0'])
# plt.yticks([300.0, 500.0], ['300', '500'])

plt.legend(loc='best')
plt.tight_layout() # 防止标签被裁切
plt.show()
```
